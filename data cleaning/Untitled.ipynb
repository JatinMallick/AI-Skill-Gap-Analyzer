```python
import pandas as pd
```

```python
df=pd.read_csv("IT_Job_Roles_Skills.csv")
```

```python
df.head()
```

```python
df=df.dropna(subset=["Skills"])
```

```python
df["Skills"] = df["Skills"].str.lower()
df["Job Description"] = df["Job Description"].str.lower()
df["Job Title"] = df["Job Title"].str.lower()
```

```python
df=df.drop("Certifications",axis=1)
df=df.drop_duplicates()
```

```python
df.describe()
```

```python
df["Job Title"].value_counts()
```

```python
df["combined_text"] = (
    df["Job Title"] + " " +
    df["Skills"] + " " +
    df["Job Description"]
)
```

```python
df.head()
```

```python
df.to_csv("dataset.csv")
```
